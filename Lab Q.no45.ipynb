```python
# Import required libraries
import numpy as np
import random

# Define states
states = ["Sunny", "Rainy"]

# Historical weather sequence
sequence = [
    "Sunny", "Sunny", "Rainy",
    "Sunny", "Rainy", "Rainy",
    "Sunny", "Sunny", "Rainy",
    "Sunny", "Sunny", "Sunny",
    "Rainy", "Rainy", "Sunny"
]

# Initialize transition matrix
transition_matrix = np.zeros((2, 2))

# Count state transitions
for i in range(len(sequence) - 1):

    current = states.index(sequence[i])
    next_state = states.index(sequence[i + 1])

    transition_matrix[current][next_state] += 1

# Calculate transition probabilities
for i in range(len(states)):

    total = np.sum(transition_matrix[i])

    if total > 0:
        transition_matrix[i] /= total

# Display transition matrix
print("Transition Probability Matrix:")
print(transition_matrix)

# Predict next event
current_state = "Sunny"

index = states.index(current_state)

probabilities = transition_matrix[index]

predicted_state = states[np.argmax(probabilities)]

print("\nCurrent Weather:", current_state)

print("Transition Probabilities:")
for i in range(len(states)):
    print(states[i], ":", probabilities[i])

print("\nPredicted Next Weather:", predicted_state)

# Generate a random next event
random_prediction = random.choices(
    states,
    weights=probabilities,
    k=1
)

print("Random Prediction:", random_prediction[0])
```
