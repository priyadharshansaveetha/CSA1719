```python
# Import required library
import numpy as np
import random

# Environment settings
states = 10
actions = 2

# Action 0 = Left, Action 1 = Right
goal = 9

# Initialize Q-table
Q = np.zeros((states, actions))

# Learning parameters
alpha = 0.1
gamma = 0.9
epsilon = 0.2
episodes = 1000

# Training the agent
for episode in range(episodes):

    state = 0
    done = False

    while not done:

        # Choose action using epsilon-greedy
        if random.random() < epsilon:
            action = random.randint(0, 1)
        else:
            action = np.argmax(Q[state])

        # Perform action
        if action == 0:
            next_state = max(0, state - 1)
        else:
            next_state = min(9, state + 1)

        # Calculate reward
        if next_state == goal:
            reward = 10
            done = True
        else:
            reward = -1

        # Update Q-value
        Q[state, action] = Q[state, action] + alpha * (
            reward + gamma * np.max(Q[next_state])
            - Q[state, action]
        )

        state = next_state

# Display Q-table
print("Trained Q-Table:")
print(Q.round(2))

# Test the trained agent
state = 0
path = [state]

while state != goal:

    action = np.argmax(Q[state])

    if action == 0:
        state = max(0, state - 1)
    else:
        state = min(9, state + 1)

    path.append(state)

print("\nAgent's Path:")
print(path)

print("\nGoal Reached Successfully!")
```
