```python
# Import required libraries
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Create user-item rating dataset
ratings = {
    'Movie1': [5, 4, 0, 0, 3],
    'Movie2': [4, 5, 0, 0, 2],
    'Movie3': [0, 0, 5, 4, 0],
    'Movie4': [0, 0, 4, 5, 0],
    'Movie5': [3, 4, 0, 0, 5]
}

# Create DataFrame
df = pd.DataFrame(
    ratings,
    index=['User1', 'User2', 'User3', 'User4', 'User5']
)

print("User-Item Rating Matrix:")
print(df)

# Calculate user similarity
similarity = cosine_similarity(df)

similarity_df = pd.DataFrame(
    similarity,
    index=df.index,
    columns=df.index
)

print("\nUser Similarity Matrix:")
print(similarity_df.round(2))

# Select target user
target_user = 'User1'

# Get similarity scores
similar_users = similarity_df[target_user].drop(
    target_user
).sort_values(ascending=False)

print("\nSimilar Users:")
print(similar_users)

# Calculate recommendations
user_ratings = df.loc[target_user]

recommendations = {}

for movie in df.columns:

    # Recommend only unrated movies
    if user_ratings[movie] == 0:

        numerator = 0
        denominator = 0

        for user, similarity_score in similar_users.items():

            rating = df.loc[user, movie]

            if rating > 0:
                numerator += similarity_score * rating
                denominator += abs(similarity_score)

        if denominator > 0:
            recommendations[movie] = (
                numerator / denominator
            )

# Display recommendations
print("\nRecommended Movies:")

for movie, score in sorted(
    recommendations.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(movie, "Predicted Rating:", round(score, 2))
```
