```python
!pip install pgmpy

# Import required libraries
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ('Rain', 'WetGrass'),
    ('Sprinkler', 'WetGrass')
])

# Define probability of Rain
cpd_rain = TabularCPD(
    variable='Rain',
    variable_card=2,
    values=[[0.7], [0.3]]
)

# Define probability of Sprinkler
cpd_sprinkler = TabularCPD(
    variable='Sprinkler',
    variable_card=2,
    values=[[0.8], [0.2]]
)

# Define probability of Wet Grass
cpd_wetgrass = TabularCPD(
    variable='WetGrass',
    variable_card=2,
    values=[
        [0.99, 0.10, 0.10, 0.01],
        [0.01, 0.90, 0.90, 0.99]
    ],
    evidence=['Rain', 'Sprinkler'],
    evidence_card=[2, 2]
)

# Add probability distributions to the network
model.add_cpds(
    cpd_rain,
    cpd_sprinkler,
    cpd_wetgrass
)

# Check model validity
print("Model is valid:", model.check_model())

# Perform probabilistic inference
inference = VariableElimination(model)

# Query probability of Wet Grass
result = inference.query(
    variables=['WetGrass']
)

print("\nProbability of Wet Grass:")
print(result)

# Query probability of Rain given Wet Grass
result2 = inference.query(
    variables=['Rain'],
    evidence={'WetGrass': 1}
)

print("\nProbability of Rain given Wet Grass:")
print(result2)
```
