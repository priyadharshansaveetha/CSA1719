```python
import random

# Genetic Algorithm parameters
POPULATION_SIZE = 10
GENERATIONS = 20
MUTATION_RATE = 0.1
CHROMOSOME_LENGTH = 5

# Fitness function
def fitness(x):
    return x ** 2

# Generate random chromosome
def create_chromosome():
    return [random.randint(0, 1)
            for _ in range(CHROMOSOME_LENGTH)]

# Convert binary chromosome to decimal
def decode(chromosome):
    value = 0
    for bit in chromosome:
        value = value * 2 + bit
    return value

# Create initial population
population = [
    create_chromosome()
    for _ in range(POPULATION_SIZE)
]

# Genetic Algorithm
for generation in range(GENERATIONS):

    # Evaluate fitness
    population.sort(
        key=lambda c: fitness(decode(c)),
        reverse=True
    )

    # Display best solution
    best = population[0]
    best_x = decode(best)

    print("Generation:", generation + 1,
          "Best X:", best_x,
          "Fitness:", fitness(best_x))

    # Selection: choose best half
    parents = population[:POPULATION_SIZE // 2]

    # Create new population
    new_population = parents.copy()

    while len(new_population) < POPULATION_SIZE:

        # Select two parents
        p1, p2 = random.sample(parents, 2)

        # Crossover
        point = random.randint(
            1, CHROMOSOME_LENGTH - 1
        )

        child = p1[:point] + p2[point:]

        # Mutation
        for i in range(CHROMOSOME_LENGTH):
            if random.random() < MUTATION_RATE:
                child[i] = 1 - child[i]

        new_population.append(child)

    population = new_population

# Final best solution
best = max(population,
           key=lambda c: fitness(decode(c)))

best_x = decode(best)

print("\nOptimal Solution:", best_x)
print("Maximum Fitness:", fitness(best_x))
```
