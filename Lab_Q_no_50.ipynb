```python

import numpy as np
import random

# XOR input and output dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])

# Neural network architecture: 2-4-1
input_size = 2
hidden_size = 4
output_size = 1

# Genetic Algorithm parameters
population_size = 50
generations = 100
mutation_rate = 0.1

# Total number of weights and biases
gene_count = (
    input_size * hidden_size +
    hidden_size +
    hidden_size * output_size +
    output_size
)

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

# Neural network prediction
def neural_network(individual, X):

    idx = 0

    # Input to hidden layer weights
    W1 = individual[idx:idx+8].reshape(2, 4)
    idx += 8

    # Hidden layer bias
    b1 = individual[idx:idx+4]
    idx += 4

    # Hidden to output weights
    W2 = individual[idx:idx+4].reshape(4, 1)
    idx += 4

    # Output bias
    b2 = individual[idx:idx+1]

    hidden = sigmoid(np.dot(X, W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)

    return output.flatten()

# Fitness function
def fitness(individual):

    predictions = neural_network(individual, X)

    error = np.mean((y - predictions) ** 2)

    return 1 / (1 + error)

# Initialize population
population = [
    np.random.uniform(-1, 1, gene_count)
    for _ in range(population_size)
]

# Selection
def selection(population):

    selected = random.sample(population, 3)

    return max(selected, key=fitness)

# Crossover
def crossover(parent1, parent2):

    point = random.randint(1, gene_count - 1)

    child = np.concatenate((
        parent1[:point],
        parent2[point:]
    ))

    return child

# Mutation
def mutation(individual):

    for i in range(gene_count):

        if random.random() < mutation_rate:
            individual[i] += np.random.normal(0, 0.5)

    return individual

# Genetic Algorithm
for generation in range(generations):

    new_population = []

    # Keep best individual (elitism)
    best = max(population, key=fitness)
    new_population.append(best.copy())

    while len(new_population) < population_size:

        parent1 = selection(population)
        parent2 = selection(population)

        child = crossover(parent1, parent2)

        child = mutation(child)

        new_population.append(child)

    population = new_population

    if generation % 10 == 0:
        best_fitness = fitness(best)
        print(
            "Generation:", generation,
            "Fitness:", round(best_fitness, 4)
        )

# Best evolved neural network
best_individual = max(population, key=fitness)

# Final predictions
predictions = neural_network(best_individual, X)

print("\nFinal XOR Predictions:")

for i in range(len(X)):
    print(
        X[i],
        "Predicted:", round(predictions[i], 3),
        "Actual:", y[i]
    )

print("\nFinal Fitness:", round(fitness(best_individual), 4))
```
